```python
from sklearn.model_selection import train_test_split
X = df.drop('target', axis = 1)
y =df['target']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.20, random_state= 42)
```

```python
# Load libraries
import numpy as np
from sklearn import datasets
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.model_selection import train_test_split 
from sklearn import svm
from sklearn.neighbors import KNeighborsClassifier
```

GridSearchCV - PIPELINE

```python
#Definición del Pipeline BAse
pipe = Pipeline(steps=[("scaler", StandardScaler()),#escalador
    ('classifier', RandomForestClassifier()) # modelo
])
```

```python
# modelos de clasificación:
#.primero logistic regresion 
logistic_reg_params = {
    'classifier': [LogisticRegression(max_iter=1000, solver='liblinear'), LogisticRegression(max_iter=10, solver='liblinear')],
    'classifier__penalty': ['l1', 'l2']
}
#segundo: random forest
random_forest_params = {
    'scaler': [StandardScaler(), MinMaxScaler()],
    'classifier': [RandomForestClassifier()],
    'classifier__max_depth': [2,3]
}
#tercero: svm
svm_param = {
    'classifier': [svm.SVC()], # kernel rbf (gaussiano)
    'classifier__C': [0.001, 0.1, 0.5, 1, 5, 10, 100],
}

#cuarto: KNN
knn_params = {
    'scaler': [StandardScaler(), MinMaxScaler()],
    'classifier': [KNeighborsClassifier()],
    'classifier__n_neighbors': [3, 5, 7, 9, 11], # Número de vecinos]
    'classifier__weights': ['uniform', 'distance'], # Ponderación de los puntos
}

#quinto:GradientBoost
gb_params = {
    'classifier': [GradientBoostingClassifier()],
    'classifier__n_estimators': [50, 100, 200], # Número de árboles
    'classifier__learning_rate': [0.01, 0.1, 0.2], 
    'classifier__max_depth': [3, 5, 7], # Profundidad de los árboles
}
```

```python
# Guardo tods los modelos anteriores en una variable
search_space = [
    logistic_reg_params,
    random_forest_params,
    svm_param,
    knn_params,
    gb_params   
]

# Entreno el modelo GrIdSearchCV

clf = GridSearchCV(estimator = pipe,
                  param_grid = search_space,
                  cv = 5,
                  n_jobs=-1)

clf.fit(X_train, y_train) # en el fit prueba todo lo anterior y devuelve el mejor modelo y los mejores parametros.
```

```python
print(clf.best_estimator_)
print(clf.best_score_)
print(clf.best_params_)
```

```python
# Hacemos la predicción para x_test
clf.best_estimator_.predict(X_test)
```

```python
clf.best_estimator_.score(X_test,y_test)
```

RandomizedSearchCV

```python
# DEFINIMOS EL MODELO
random_search = RandomizedSearchCV(
    estimator = pipe,
    param_distributions = search_space, # Aquí se llama param_distributions en lugar de param_grid
    n_iter = 50,    # Número de combinaciones a probar
    cv = 5,         # Validación cruzada
    verbose = 1,    # Para que vaya contando qué hace
    n_jobs = -1,    # Para usar todos MIs procesadores(CPU)
    random_state = 42 # Para que los resultados sean reproducibles
)

# Entrenamos
result = random_search.fit(X_train, y_train)
```

```python
print(random_search.best_estimator_)
print(random_search.best_params_)
print(random_search.best_score_)
```

```python
# Predicción
y_pred = random_search.predict(X_test)
y_pred
```

```python

from sklearn.metrics import classification_report, confusion_matrix
print(classification_report(y_test, y_pred))
```

```python
# Guardamos el modelo 
import pickle # libreria de python para guardar objetos

#me guardo el modelo 
filename = 'finished_model'

with open(filename, 'wb') as archivo_salida:
    pickle.dump(random_search.best_estimator_, archivo_salida)
```
