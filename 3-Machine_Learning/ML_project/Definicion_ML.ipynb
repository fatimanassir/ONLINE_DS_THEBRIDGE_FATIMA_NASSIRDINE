### TEMA

## Tema del proyecto

El presente proyecto se centra en la predicción de cancelaciones de reservas hoteleras a partir de datos históricos. En concreto, se pretende analizar si una reserva ya realizada será finalmente cancelada o no, utilizando la información disponible en el momento de la reserva.

Para ello, se estudiarán distintos factores relacionados con el cliente y las características de la reserva, como la antelación, el tipo de depósito o el historial previo, con el objetivo de identificar patrones de comportamiento asociados a la cancelación.

Este enfoque permite no solo construir un modelo predictivo, sino también comprender mejor el comportamiento de los clientes, aportando información útil para la toma de decisiones en la gestión hotelera.

### OBJETIVOS

Define aquí el objetivo del proyecto

Objetivo Principal:<br>
Desarrollar un modelo de aprendizaje supervisado (clasificación) capaz de predecir con alta precisión si una reserva de hotel será cancelada (is_canceled = 1) o no (is_canceled = 0), basándose en la información disponible en el momento de la creación de la reserva.<br>

Objetivos Secundarios:<br>

Identificar las variables clave (como el lead_time, deposit_type o total_of_special_requests) que tienen mayor peso en las cancelaciones.<br>

Proporcionar una herramienta que permita a los hoteles optimizar su gestión de inventario y reducir las pérdidas por habitaciones vacías.<br>
