```python
# Nos quedamos con rating (numérica), y las categóricas de menos de 10 categorias.
df_model = df[['rating', 'effectiveness', 'sideEffects']]
df_model
```

#### Transforma las columnas categóricas

Transforma las columnas categoricas a numericas mediante dummies

```python
# Transformamos a dummies con get.dummies
df_model = pd.get_dummies(df_model, columns=['effectiveness', 'sideEffects'])
df_model. info()
```

#### Evalua cual es la mejor K

Utiliza silhouette_score para evaluar cual es la mejor K.

#### Genera el K Means 

```python
from sklearn.cluster import KMeans
X = df_model

# Definimos K_meas, el numero de grupos 
k = 5 # 5 grupos de cosas

# definimos el modelo
kmeans = KMeans(n_clusters=k, n_init=10, random_state=42) 

y_pred = kmeans.fit_predict(X)
#en este modelo no se separa train y test. poruqe no tengo target y no tengo con que compararlo :)

y_pred
```

```python
len(y_pred)
```

```python
np.unique(y_pred)
```

```python
# Añadimos la predicción al dataframe
df_model['cluster'] = y_pred

# Echamos un vistazo para ver cómo ha quedado
df_model.head()
```

```python
# Contamos cuántos elementos hay en cada cluster
distribucion = df_model['cluster'].value_counts()
```
