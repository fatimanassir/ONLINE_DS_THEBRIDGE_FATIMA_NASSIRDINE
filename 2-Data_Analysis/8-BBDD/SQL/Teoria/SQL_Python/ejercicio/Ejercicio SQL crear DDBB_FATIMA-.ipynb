```python
# Función para ejecutar una consulta SQL y devolver un DataFrame
def sql_query(query, cursor):
    """
    Ejecuta una query SQL usando el cursor proporcionado
    y devuelve los resultados como un DataFrame de Pandas.

    Parámetros:
    - query : str : La consulta SQL a ejecutar
    - cursor : sqlite3.Cursor : Cursor de la conexión a la base de datos

    Retorna:
    - pd.DataFrame con los resultados de la consulta
    """
    # Ejecutamos la consulta SQL
    cursor.execute(query)

    # Guardamos los resultados en una lista de tuplas
    resultados = cursor.fetchall()

    # Obtenemos los nombres de las columnas de la consulta
    columnas = [descripcion[0] for descripcion in cursor.description]

    # Convertimos los resultados a un DataFrame de Pandas y devolvemos
    return pd.DataFrame(resultados, columns=columnas)
```

1. Obtener el nombre de los empleados que hayan repartido al bar **Stop** durante la semana del 17 al 23 de octubre de 2005.

```python
#Ejecutar consulta
cursor.execute("""
SELECT DISTINCT EMPLEADOS.Nombre
FROM REPARTO
JOIN EMPLEADOS ON REPARTO.CodE = EMPLEADOS.CodE
JOIN BARES ON REPARTO.CodB = BARES.CodB
WHERE BARES.Nombre = 'Stop'
    AND REPARTO.Fecha BETWEEN '17/10/05' AND '23/10/05';
""")

# Aplico la funcion
df_empleados_stop = sql_query("""
SELECT DISTINCT EMPLEADOS.Nombre
FROM REPARTO
JOIN EMPLEADOS ON REPARTO.CodE = EMPLEADOS.CodE ---JOIN para conectar las tablas con sus claves.
JOIN BARES ON REPARTO.CodB = BARES.CodB
WHERE BARES.Nombre = 'Stop'
  AND REPARTO.Fecha BETWEEN '10/17/05' AND '10/23/05';
""", cursor)

# Mostramos el resultado
df_empleados_stop
```

2. Obtener el Cif y nombre de los bares a los que se ha repartido cerveza de tipo **Botella** y capacidad inferior a 1 litro, ordenados por localidad.

```python
#Ejecutar consulta
cursor. execute("""
                SELECT DISTINCT BARES.Cif, BARES.Nombre ---DISTINCT para que no repita bares si tienen varios repartos que cumplen la condición.
                FROM REPARTO
                JOIN BARES ON REPARTO.CodB = BARES.CodB ---JOIN para conectar las tablas con sus claves.
                JOIN CERVEZAS ON REPARTO.CodC = Cervezas.CodC
                WHERE CERVEZAS.Envase = 'Botella'
                    AND CERVEZAS.Capacidad < 1
                ORDER BY BARES.Localidad;
                """)
# Aplico la funcion
df_bares_botella = sql_query("""
                SELECT DISTINCT BARES.Cif, BARES.Nombre ---DISTINCT para que no repita bares si tienen varios repartos que cumplen la condición.
                FROM REPARTO
                JOIN BARES ON REPARTO.CodB = BARES.CodB ---JOIN para conectar las tablas con sus claves.
                JOIN CERVEZAS ON REPARTO.CodC = Cervezas.CodC
                WHERE CERVEZAS.Envase = 'Botella'
                    AND CERVEZAS.Capacidad < 1
                ORDER BY BARES.Localidad;
                """, cursor)

# Mostramos el resultado
df_bares_botella
```

3. Obtener los repartos (nombre del bar, envase y capacidad de la bebida, fecha y cantidad) realizados por **Prudencio Caminero**.

```python
#construiR la consulta:
"""
SELECT BARES.Nombre, CERVEZAS.Envase, CERVEZAS.Capacidad, REPARTO.Fecha, REPARTO.Cantidad
FROM REPARTO
JOIN EMPLEADOS ON REPARTO.CodE = EMPLEADOS.CodE
JOIN BARES ON REPARTO.CodB = BARES.CodB
JOIN CERVEZAS ON REPARTO.CodC = CERVEZAS.CodC
WHERE EMPLEADOS.Nombre = 'Prudencio Caminero';

"""
# Aplico la funcion
df_repartos_prudencio = sql_query("""
                SELECT BARES.Nombre, CERVEZAS.Envase, CERVEZAS.Capacidad, REPARTO.Fecha, REPARTO.Cantidad
                FROM REPARTO
                JOIN EMPLEADOS ON REPARTO.CodE = EMPLEADOS.CodE
                JOIN BARES ON REPARTO.CodB = BARES.CodB
                JOIN CERVEZAS ON REPARTO.CodC = CERVEZAS.CodC
                WHERE EMPLEADOS.Nombre = 'Prudencio Caminero';

                """, cursor)

# Mostramos el resultado
df_repartos_prudencio
```

4. Obtener los bares a los que se les ha repartido envases de tipo **botella** y capacidad 0.2 ó 0.33.

```python
#construiR la consulta:
"""
SELECT BARES.Nombre
FROM REPARTO
JOIN BARES ON REPARTO.CodB = BARES.CodB
JOIN CERVEZAS ON REPARTO.CodC = CERVEZAS.CodC
WHERE CERVEZAS.Envase = 'Botella'
    AND CERVEZAS.Capacidad IN (0.2, 0.33)
ORDER BY BARES.Nombre;
    
"""
# Aplico la funcion
df_bares_botella_peq = sql_query("""
                SELECT BARES.Nombre
                FROM REPARTO
                JOIN BARES ON REPARTO.CodB = BARES.CodB
                JOIN CERVEZAS ON REPARTO.CodC = CERVEZAS.CodC
                WHERE CERVEZAS.Envase = 'Botella'
                    AND CERVEZAS.Capacidad IN (0.2, 0.33)
                ORDER BY BARES.Nombre;
                """, cursor)

# Mostramos el resultado
df_bares_botella_peq
```

5. Nombre de los empleados que han repartido a los bares **"Stop"** y **"Las Vegas"** cervezas con envase botella. 

```python
#construiR la consulta:
"""
SELECT EMPLEADOS.Nombre
FROM REPARTO
JOIN BARES ON REPARTO.CodB = BARES.CodB
JOIN CERVEZAS ON REPARTO.CodC = CERVEZAS.CodC
JOIN EMPLEADOS ON REPARTO.CodE = EMPLEADOS.CodE
WHERE BARES.Nombre IN ('Stop', 'Vegas')
    AND CERVEZAS.Envase = 'Botella'
GROUP BY EMPREADOS.Nombre;
    
"""
# Aplico la funcion
df_empleados_stop_vegas = sql_query("""
                SELECT EMPLEADOS.Nombre
                FROM REPARTO
                JOIN BARES ON REPARTO.CodB = BARES.CodB
                JOIN CERVEZAS ON REPARTO.CodC = CERVEZAS.CodC
                JOIN EMPLEADOS ON REPARTO.CodE = EMPLEADOS.CodE
                WHERE BARES.Nombre IN ('Stop', 'Vegas')
                    AND CERVEZAS.Envase = 'Botella'
                GROUP BY EMPLEADOS.Nombre;
                """, cursor)

# Mostramos el resultado
df_empleados_stop_vegas
```

6. Obtener el nombre y número de viajes que ha realizado cada empleado fuera de **Villa Botijo**.

```python
#construiR la consulta:
"""
SELECT EMPLEADO.Nombre, COUNT(*) AS NumeroViajes
FROM REPARTO
JOIN EMPLEADO ON REPARTO.CodE =EMPLEADO.CodE
JOIN BARES ON REPARTO.CodB = BARES.CodB
WHERE BARES.Localidad != 'Villa Botijo'
GROUP BY EMPLEADOS.Nombre
ORDER BY EMPLEADOS.Nombre;
"""

# Aplico la funcion
df_viajes_fuera  = sql_query("""
                SELECT EMPLEADOS.Nombre, COUNT(*) AS NumeroViajes
                FROM REPARTO
                JOIN EMPLEADOS ON REPARTO.CodE =EMPLEADOS.CodE
                JOIN BARES ON REPARTO.CodB = BARES.CodB
                WHERE BARES.Localidad != 'Villa Botijo'
                GROUP BY EMPLEADOS.Nombre
                ORDER BY EMPLEADOS.Nombre;
                """, cursor)
# Mostramos el resultado
df_viajes_fuera
```

7. Obtener el nombre y localidad del bar que más litros de cerveza ha comprado.

```python
#construiR la consulta:
"""
SELECT BARES.Nombre, BARES.Localidad, SUM(REPARTO.Cantidad * CERVEZAS.Capacidad) AS LitrosTotales
FROM REPARTO
JOIN BARES ON REPARTO.CodB = BARES.CodB
JOIN CERVEZAS ON REPARTO.CodC = CERVEZAS.CodC
GROUP BY BARES.Nombre, BARES.Localidad
"""
# Aplico la funcion
df_bar_max_litros = sql_query("""
SELECT BARES.Nombre, BARES.Localidad, SUM(REPARTO.Cantidad * CERVEZAS.Capacidad) AS LitrosTotales
FROM REPARTO
JOIN BARES ON REPARTO.CodB = BARES.CodB
JOIN CERVEZAS ON REPARTO.CodC = CERVEZAS.CodC
GROUP BY BARES.Nombre, BARES.Localidad
""", cursor)
# Mostramos el resultado
df_bar_max_litros
```

8. Obtener los bares que han adquirido todos los tipos de cerveza con envase de botella y capacidad menor que 1 litro.

```python
#construiR la consulta:
"""
SELECT BARES.Nombre
FROM REPARTO
JOIN BARES ON REPARTO.CodB = Bares.CodB
JOIN CERVEZAS ON REPARTO.CodC = CERVEZAS.CodC
WHERE CERVEZAS.Envase = 'Botella'
    AND CERVEZAS.Capacidad < 1
GROUP BY BARES.Nombre
HAVING COUNT(DISTINCT CERVEZAS.Capacidad) = 2; 
"""
"""HAVING COUNT(DISTINCT CERVEZAS.Capacidad) = 2 #HAVING: filtra los grupos (después del GROUP BY),COUNT(...): cuenta cuántos valores hay,2: queremos solo los bares que tengan 2 capacidades distintas"""

# Aplico la funcion
df_bares_completos = sql_query("""
                              SELECT BARES.Nombre
                                FROM REPARTO
                                JOIN BARES ON REPARTO.CodB = Bares.CodB
                                JOIN CERVEZAS ON REPARTO.CodC = CERVEZAS.CodC
                                WHERE CERVEZAS.Envase = 'Botella'
                                    AND CERVEZAS.Capacidad < 1
                                GROUP BY BARES.Nombre
                                HAVING COUNT(DISTINCT CERVEZAS.Capacidad) = 2; 
                                """, cursor)
# Mostramos el resultado
df_bares_completos
```

9. Subir un 5% el sueldo del empleado que más días haya trabajado.

```python
#construiR la consulta:
#PRIMERO ENCONTRAR EMPLEADO QUE MAS HA TRABAJADO

"""
SELECT CodE
FROM REPARTO
GROUP BY CodE --- agrupa por empleado
Order by COUNT(*) DESC --- cuénta cuantos repartos hizo
LIMIT 1; --- el que más tiene
"""
#segundo subir el sueldo, aqui mi función sql_query() NO sirve para UPDATE. solo funciona con consultas SELECT
cursor.execute("""
            UPDATE EMPLEADOS
            SET Sueldo = Sueldo * 1.05
            WHERE CodE =(SELECT CodE
                        FROM REPARTO
                        GROUP BY CodE --- agrupa por empleado
                        Order by COUNT(*) DESC --- cuénta cuantos repartos hizo
                        LIMIT 1);--- el que más tiene
            """)
                        
connect.commit()

#comprobar el Update

df_empleados = sql_query("SELECT * FROM EMPLEADOS", cursor)
df_empleados 
```
