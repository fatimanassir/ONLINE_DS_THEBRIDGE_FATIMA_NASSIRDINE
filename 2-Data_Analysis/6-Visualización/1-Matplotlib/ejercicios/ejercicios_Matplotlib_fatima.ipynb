2. Sabemos que podemos pintar gráficas de dos formas: la figura incluye los ejes o teniendo figura y ejes por separado.

Usando solo una figura, usa numpy para los valores del eje X entre 0 y 5. 

Pinta dos gráficas en dos cajas distintas, a la izquierda una recta con pendiente positiva de 3 que pase por (0,0) y a la derecha una recta con pendiente negativa de 3 que pase por (0,-5). 

Elige la precisión en el eje X que desees.  

```python
x = np.linspace(0, 5, 100) # creo 100 puntos enter 0 y 5

plt.figure(figsize=(10,4)) # figura con dos cajas

# gráfica izq:
plt.subplot(1,2,1) # 1fila, 2 cols, posición 1.
plt.plot(x, 3*x)   # pendiente +3, pasa por (0,0)

# gráfica dcha:
plt.subplot(1,2,2)
plt.plot(x, -3*x - 5)  # pendiente -3, pasa por (0,-5)

plt.show()
```

3. Fija el eje X entre 0 y 5 y el eje Y entre -15 y 15

```python
plt.figure(figsize=(10,4)) #figsize=(10,4) defime el tamaño de la figura.
plt.subplot(1, 2, 1)
plt.plot(x, 3*x)
plt.xlim(0,5) #límites eje x
plt.ylim(-15,15) # límites eje y

plt.subplot(1, 2, 2)
plt.plot( x, -3*x -5)
plt.xlim(0,5)
plt.ylim(-15,15)

plt.show()
```

```python
# x = np.linspace(0, 5, 100) # creo 100 valores enter 0 y 5

plt.figure(figsize=(10,4)) # figura con dos cajas

# gráfica izq:
plt.subplot(1,2,1) # 1fila, 2 cols, posición 1
plt.plot(x, 3*x)   # pendiente +3, pasa por (0,0)
plt.title("recta sube") # título
plt.xlabel("eje X")
plt.ylabel("eje y")

# gráfica dcha:
plt.subplot(1,2,2)
plt.plot(x, -3*x - 5)  # pendiente -3, pasa por (0,-5)
plt.title("recta baja") # título
plt.xlabel("eje X") # solo quiero etiqueta x aquí
plt.show()
```

13. Vamos a introducir pandas. Crea un DataFrame con las columnas "Year" de valores 2015, 2016, 2017, 2018, 2019 y la columna "Sold_items_A" de valores 1000, 3500, 4000, 5500, 7000

```python
data =  pd.DataFrame({"Year":[2015, 2016, 2017, 2018, 2019], "Sold_items_A": [1000, 3500, 4000, 5500, 7000]})
```

14. Ahora pintamos un gráfico de línea con las ventas respecto al año. Pon un título y etiquetas en los ejes.
Cambia el estilo a 'seaborn-white'. La línea debe ser con rayas y verde.

```python
plt.plot(data["Year"], data["Sold_items_A"], 
         linestyle="--", #me genera linea discontinua
         color = "green")
plt.title("Ventas departamento A")
plt.xlabel("Año")
plt.ylabel("Unidades vendidas")

plt.show()
```

15. Otro departamento B ha vendido en esos años 2000, 3100, 5000, 4000, 6000 unidades. Incluye esa columna en el DataFrame y pinta en la misma gráfica las dos líneas. B es una línea punteada y roja. Muestra una leyenda abajo a la derecha.

```python
data["Sold_items_B"] = [2000, 3100, 5000, 4000, 6000]

plt.plot(data["Year"], data["Sold_items_A"], linestyle="--", color = "green", label="Dept A")

plt.plot(data["Year"], data["Sold_items_B"], linestyle=":", color="red", label="Dept B")

plt.legend(loc="lower right") # leyenda abajo a la derecha
plt.title("Ventas por departamento")
plt.xlabel("Año")
plt.ylabel("Unidades vendidas")

plt.show()
```

19. Vamos a pintar un histograma de unas alturas generadas aleatoriamente. Crea 300 alturas entre 150 y 190 cm.

Pista: https://pandas.pydata.org/pandas-docs/version/0.23.4/generated/pandas.DataFrame.plot.html

```python
alturas = np.random.randint(150, 191, 300) # 300 alturas entre 150 y 190 cm

plt.hist(alturas, bins=15) # histograma
plt.xlabel("Altura (cm)")
plt.ylabel("Frecuencia")

plt.show()
```
