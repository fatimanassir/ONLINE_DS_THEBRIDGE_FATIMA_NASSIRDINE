## Ejercicio 1
Escribe una función que convierta números del 1 al 7 en nombres de los dias de la semana. La función constará de un único argumento numérico y una salida de tipo string

````python
def dias_semana(numero):
    dias = ["lunes", "martes", "miercoles", "jueves", "viernes","sabado", "domingo"]
    if 1 <= numero <= 7:
        return dias[numero - 1]
    
````

````python
print(dias_semana(1))
print(dias_semana(2))
print(dias_semana(3))
print(dias_semana(4))
print(dias_semana(5))
print(dias_semana(6))
print(dias_semana(7))
````

## Ejercicio 3
Escribe una función que compare dos números. La función tiene dos argumentos y hay tres salidas posibles: que sean iguales, que el primero sea  mayor que el segundo, o que el segundo sea mayor que el primero

````python
def comparacion(num1, num2):

    if num1 == num2:
        return f"los numeros,{num1} y {num2} son iguales"
    elif num1 > num2:
            return f"el numero {num1} es mayor que el numero {num2}"
    elif num1 < num2:
        return f"el numero {num2} es mayor que {num1}"
 
    

````

````python
print(comparacion(2, 2))
print(comparacion(2, 19))
print(comparacion(-10, 4))
````

## Ejercicio 4
Escribe una función que sea un contador de letras. En el primer argumento tienes que introducir un texto, y el segundo que sea una letra que contar. La función tiene que devolver un entero con el número de veces que aparece esa letra, tanto mayúscula, como minúscula

````python
def num_letras(text, letra):
    return text.lower().count(letra.lower())
        
````

````python
print(num_letras("Me siento perdida", "M"))
print(num_letras("Me siento perdida", "s"))
print(num_letras("Me siento perdida", "e"))
print(num_letras("Me siento perdida", "d"))
````
