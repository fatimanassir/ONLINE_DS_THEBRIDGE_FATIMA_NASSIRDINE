### 1. Utiliza map para elevar al cuadrado cada número en la lista [1, 2, 3, 4, 5].

Solución esperada: [1, 4, 9, 16, 25]

### 2. Filtrar números pares de una lista:

Emplea filter para obtener sólo los números pares de la lista [1, 2, 3, 4, 5, 6].

Solución esperada: [2, 4, 6]

```python
lista1= [1, 2, 3, 4, 5, 6]
lista2= filter(lambda x: x % 2 == 0, lista1)
print(list(lista2))
```

### 3. Convertir grados Celsius a Fahrenheit:

Usa map y una función lambda para convertir una lista de temperaturas en grados Celsius [0, 10, 20, 30] a Fahrenheit.

Solución esperada: [32.0, 50.0, 68.0, 86.0]

```python
grados_celsius = [0, 10, 20, 30]
grados_fahrenheit = (map(lambda x: ((x *9) / 5) + 32, grados_celsius))
print(list(grados_fahrenheit))
        
```

### 4. Sumar los cuadrados de los números pares de una lista:

Combina filter, map y reduce (importa reduce desde functools) para sumar los cuadrados de los números pares en [1, 2, 3, 4, 5, 6].

Solución esperada: 56

```python
mi_lista = [1,2,3,4,5,6]
# Usamos filter para que nos devuelva aquellos elementos que son pares. es decir los que cumplen la condicion True.
func_filter= list(filter(lambda x: x % 2 == 0, mi_lista)) #guardo el resultado de la funcion_filter en una lista.porque la funcion filter se consume una vez. sino lo hubiese puesto list, al hacer el print abajo, me devolveria el output pero la func_filter quedaria vacia.
print(func_filter)

#Luego usamos map para elevar al cuadrado los elementos de la lista func_filter.
func_map = list(map(lambda x: x**2, func_filter))
print(list(func_map))

#Para la funcion reduce, priemro importamos reduce de la libreria functools.
from functools import reduce
func_reduce = reduce(lambda x,y: x +y, func_map) # en el segundo argomento introduzco la variable func_map.
print(func_reduce)
```

### 5. Encadenar map, filter, y reduce para resolver un problema complejo:

Dada una lista de strings que representan números enteros y "NA" (por ejemplo, ["2", "5", "NA", "9", "10"]), convierte los números a enteros, filtra los "NA", suma el cuadrado de los números que son mayores que 5 y menores que 10.

Solución esperada: Un único valor que es la suma de los cuadrados de los números válidos.

```python
mi_lista_string = ["1", "5", "NA", "NA","10", "13","6"]
```

```python
from functools import reduce

# Descarto los elementos de la lista que tienen "NA". para ello uso la funcion filter, y la condicion es que me devuelva una lista de todos  los elementos distintos de "NA".

mi_lista_filter= list(filter(lambda x: x != "NA", mi_lista_string)) 
print(mi_lista_filter)

# Aqui con funcion map, le estoy pidiendo que cada uno de los elementos de mi_lista_filter, me los convierta primero a float, y luego a int. para todos los elementos de la lista. 

mi_lista_map = list(map(lambda x: int(float(x)) , mi_lista_filter))
print(mi_lista_map)    

# en esta marte he puesto la condicion, si el numero esta entre el 5-10 me devuelve el cuadrado de ese numero, y sino el mismo numero. (he hecho un poco trampita porque no se como hacerlo en un solo paso.)
mi_lista_map1=list(map(lambda x: x**2 if (5 < x < 10 ) else x, mi_lista_map))
print(mi_lista_map1)
# con los resultados  de mi_lista_map1, hago la funcion reduce, que me devuelve un numero que es la suma de los elementos de mi_lista_map1.
mi_lista_reduce = reduce(lambda x,y: x + y, mi_lista_map1)
print(mi_lista_reduce)
      
```
